* 환경변수 초기화 팁

```python
import os

# 1개 삭제
# 환경변수가 있으면 삭제
# 환경변수가 없어도 에러는 발생하지 않는다.
os.environ.pop("APP_MODE", None)

# 여러개 삭제
# 지우고 싶은 환경변수명 리스트
keys = ["APP_MODE", "ANTHROPIC_API_KEY"]

for key in keys:
    os.environ.pop(key, None)
```
