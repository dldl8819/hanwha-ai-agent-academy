```python
# 대화는 이력이 쌓여야 한다.
chat: list[dict[str, str]] = []

# 말 한 마디를 대화 이력에 쌓는 함수
def append_turn(history: list[dict[str, str]], role: str, content: str) -> None:
    history.append({"role": role, "content": content})

append_turn(chat, "user", "오늘 날씨가 어때?")
append_turn(chat, "assistant", "기상청 정보를 조회할 수 없습니다.")

for message in chat:
    print(f"{message["role"]:<10}{message["content"]}")

print(f"대화 이력 길이 : {len(chat)}개")
```
